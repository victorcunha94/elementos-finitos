## 1. Geração da partição — item 1

O próximo bloco define a partição triangular estruturada. Cada retângulo da grade é dividido em dois triângulos, cujos vértices são registrados em `triangulos`. Para os parâmetros gerais, o domínio é $[0,L_x]\times[0,L_y]$; nos experimentos, $L_x=L_y=1$.
$$
N_v=n_xn_y,\qquad N_{el}=2(n_x-1)(n_y-1).
$$
`nx` e `ny` contam pontos, e não intervalos. `nodes` contém inicialmente as coordenadas dos vértices geométricos. A função apenas é definida nesta célula; quando for chamada, `numpy` já terá sido importado na célula de quadratura. Nenhuma condição de contorno é imposta nesta etapa.

```python
def gerar_malha_2d(lx = 1.0, ly = 1.0, nx = 5, ny = 5):
    """Gera uma malha estruturada de triângulos no domínio [0, 1] x [0, 1].
    """
    x = np.linspace(0, lx, nx)
    y = np.linspace(0, ly, ny)
    X, Y = np.meshgrid(x, y)
    nodes = np.column_stack((X.ravel(), Y.ravel()))

    triangulos = []
    for j in range(ny - 1):
        for i in range(nx - 1):
            n0 = j * nx + i
            n1 = n0 + 1
            n2 = n0 + nx
            n3 = n2 + 1
            # Triângulo 1
            triangulos.append([n0, n1, n2])
            # Triângulo 2
            triangulos.append([n1, n3, n2])

    return np.array(nodes), np.array(triangulos)
```

## 2. Espaços de aproximação e conectividade — itens 1 e 2

A rotina configura os dois espaços descritos na seção 8.1:
$$
X_h(\mathcal T_h)=P_1^{\mathrm{disc}}(\mathcal T_h)
=\{v\in L^2(\Omega):v|_{K_m}\in P_1(K_m)\},
\qquad
V_h(\mathcal T_h)=X_h(\mathcal T_h)\cap C^0(\overline\Omega).
$$
No código, `DG1` representa $X_h$ e `CG1` representa $V_h$. Para `CG1`, os elementos adjacentes compartilham os graus de liberdade dos vértices, e $\dim V_h=N_v$. Para `DG1`, cada triângulo tem três graus de liberdade próprios, e $\dim X_h=3N_{el}$; coordenadas coincidentes não significam a mesma incógnita.

A matriz `elements_espaco`, depois chamada `elements`, corresponde à matriz de incidência `conec` da apostila:
$$
I=LG_{K_m}(r)=\mathrm{conec}(K_m,r),\qquad
\mathrm{conec}\in\mathbb N^{N_{el}\times n_{loc}},\quad n_{loc}=3.
$$
Na apostila os índices começam em $1$; em Python começam em $0$. Assim, `elements[m,r]` fornece o índice global associado ao índice local `r` do elemento `m`.

Preservamos o nome `nodes`: antes da configuração, são vértices geométricos; depois, são as coordenadas associadas aos graus de liberdade do espaço nodal $P_1$. Os valores desconhecidos não ficam em `nodes`, mas no vetor `uh`. Em `DG1` há três entradas por elemento, inclusive quando suas coordenadas coincidem com as de elementos vizinhos.

```python
def configurar_espaco_aproximacao(nodes, triangulos, type_space="CG1"):
    """
    Configura a conectividade dos elementos e as coordenadas dos nós baseando-se
    no espaço de aproximação selecionado (CG1 ou DG1).

    Parâmetros:
    -----------
    nodes : ndarray
        Coordenadas dos vértices geométricos da malha original.
    triangulos : ndarray
        Conectividade geométrica original dos triângulos.
    type_space : str, opcional
        Tipo do espaço de aproximação ('CG1' ou 'DG1'). O padrão é 'CG1'.

    Retorna:
    --------
    nodes_espaco : ndarray
        As coordenadas físicas dos nós configurados para o espaço.
    elements_espaco : ndarray
        A tabela de conectividade (elementos) configurada para o espaço.
    """
    if type_space == "CG1":
        nodes_espaco = nodes
        elements_espaco = triangulos
    elif type_space == "DG1":
        # Para DG1: Cada elemento possui seus próprios 3 nós sem compartilhamento
        nodes_dg = []
        elements_dg = []
        no_id = 0
        for tri in triangulos:
            # Adiciona as coordenadas físicas dos 3 nós do triângulo atual
            nodes_dg.append(nodes[tri[0]])
            nodes_dg.append(nodes[tri[1]])
            nodes_dg.append(nodes[tri[2]])

            # Associa a conectividade local exclusiva para este elemento
            elements_dg.append([no_id, no_id + 1, no_id + 2])
            no_id += 3
        nodes_espaco = np.array(nodes_dg)
        elements_espaco = np.array(elements_dg)
    else:
        raise ValueError("type_space deve ser 'CG1' ou 'DG1'.")

    return nodes_espaco, elements_espaco
```

## 3. Regras de quadratura no elemento mestre — item 1

A seção 8.2 define os pontos $\widehat{\mathbf x}_g$ e pesos $w_g$, $g=1,\ldots,n_g$. Utilizamos
$$
\widehat K=\{(\xi,\eta):\xi\geq0,\ \eta\geq0,\ \xi+\eta\leq1\},\qquad |\widehat K|=\frac12.
$$
As opções `linear`, `quadratic`, `cubic` e `quintic` têm, respectivamente, $1,3,4,7$ pontos e graus de exatidão $1,2,3,5$, com a precisão dos coeficientes tabulados. A regra cúbica possui um peso negativo; portanto, sua aplicação a uma função não polinomial positiva não garante, por si só, uma soma positiva. Nas avaliações finais de erro utiliza-se a regra quíntica, de pesos positivos.

**Equivalência das coordenadas.** A função retorna triplas baricêntricas $(L_1,L_2,L_3)$, mas as rotinas seguintes usam $(\xi,\eta)=(L_1,L_2)$. Com a ordenação das bases do programa, os valores nodais correspondem a $(1-\xi-\eta,\xi,\eta)=(L_3,L_1,L_2)$. Trata-se de uma permutação; como os conjuntos de pontos e pesos empregados são simétricos, ela preserva a regra.

**Diferença de normalização em relação à definição 8.2.1.** Chamando $\widetilde w_g$ os valores de `pesos`, tem-se $\sum_g\widetilde w_g\simeq1$. Para integrar no triângulo mestre de área $1/2$, os pesos da apostila são $w_g=\widetilde w_g/2$:
$$
\int_{K_m}q(\mathbf x)\,dK
\approx |J_{K_m}|\sum_g w_g q(F_{K_m}(\widehat{\mathbf x}_g))
=\frac{|J_{K_m}|}{2}\sum_g\widetilde w_g q(F_{K_m}(\widehat{\mathbf x}_g)).
$$
O código usa $|J_{K_m}|\widetilde w_g$, sem o fator $1/2$. Logo, suas somas de integração são duas vezes as somas com a normalização da apostila. Nas montagens deste notebook, matriz e vetor são multiplicados pelo mesmo fator, que se cancela na solução do sistema. Nos erros, o fator permanece: a raiz da soma fica multiplicada por $\sqrt2$. Este relatório registra a diferença; o código permanece intacto.

```python
import numpy as np


def pontos_pesos_triangulo(ordem):
    """
    Retorna os pontos de Gauss e pesos da tabela para o triângulo.

    ordem : str
        "linear", "quadratic", "cubic" ou "quintic".

    Retorno
    -------
    pontos_gauss : ndarray, formato (nq, 3)
        Coordenadas baricêntricas (L1, L2, L3).
    pesos : ndarray, formato (nq,)
        Pesos normalizados, cuja soma é 1.
    """

    ordem = ordem.lower()

    if ordem == "linear":
        pontos_gauss = [
            (1/3, 1/3, 1/3)
        ]
        pesos = [
            1.0
        ]

    elif ordem == "quadratic":
        pontos_gauss = [
            (1/2, 1/2, 0),
            (0,   1/2, 1/2),
            (1/2, 0,   1/2)
        ]
        pesos = [
            1/3,
            1/3,
            1/3
        ]

    elif ordem == "cubic":
        pontos_gauss = [
            (1/3, 1/3, 1/3),
            (0.6, 0.2, 0.2),
            (0.2, 0.6, 0.2),
            (0.2, 0.2, 0.6)
        ]
        pesos = [
            -27/48,
            25/48,
            25/48,
            25/48
        ]

    elif ordem == "quintic":
        alpha_1 = 0.0597158717
        beta_1  = 0.4701420641
        alpha_2 = 0.7974269853
        beta_2  = 0.1012865073

        pontos_gauss = [
            (1/3, 1/3, 1/3),

            (alpha_1, beta_1, beta_1),
            (beta_1, alpha_1, beta_1),
            (beta_1, beta_1, alpha_1),

            (alpha_2, beta_2, beta_2),
            (beta_2, alpha_2, beta_2),
            (beta_2, beta_2, alpha_2)
        ]
        pesos = [
            0.2250000000,

            0.1323941527,
            0.1323941527,
            0.1323941527,

            0.1259391805,
            0.1259391805,
            0.1259391805
        ]

    else:
        raise ValueError(
            'ordem deve ser "linear", "quadratic", "cubic" ou "quintic".'
        )

    return np.array(pontos_gauss, dtype=float), np.array(pesos, dtype=float)
```

## 4. Transformação afim — item 1

Para os vértices físicos $\mathbf v_0,\mathbf v_1,\mathbf v_2$ de $K_m$, a rotina constrói
$$
F_{K_m}(\widehat{\mathbf x})=B_{K_m}\widehat{\mathbf x}+\mathbf v_0,
\qquad
B_{K_m}=\begin{bmatrix}\mathbf v_1-\mathbf v_0&\mathbf v_2-\mathbf v_0\end{bmatrix}.
$$
`getBk` retorna $B_{K_m}$. Nas rotinas de montagem, `detJac` corresponde a $|J_{K_m}|=|\det B_{K_m}|$, e `invBk_T` corresponde a $B_{K_m}^{-T}$. A área física é $|K_m|=|J_{K_m}|/2$. A transformação leva os vértices $(0,0),(1,0),(0,1)$ aos três vértices físicos, permitindo reutilizar as mesmas bases e regras de quadratura em todos os elementos.

```python

def getBk(element, nodes):
    v0 = nodes[element[0]]
    v1 = nodes[element[1]]
    v2 = nodes[element[2]]
    return np.array([[v1[0] - v0[0], v2[0] - v0[0]], [v1[1] - v0[1], v2[1] - v0[1]]])
```

## 5. Função a projetar, bases locais e gradientes — itens 1 e 2

Escolhe-se a função suave
$$
u(x,y)=\sin(2\pi x)+\frac12\cos(5\pi y).
$$
As três funções de forma do elemento mestre são
$$
\widehat\psi_{\widehat K}^1=1-\xi-\eta,\qquad
\widehat\psi_{\widehat K}^2=\xi,\qquad
\widehat\psi_{\widehat K}^3=\eta,
$$
$$
\widehat\nabla\widehat\psi_{\widehat K}^1=\begin{bmatrix}-1\\-1\end{bmatrix},\quad
\widehat\nabla\widehat\psi_{\widehat K}^2=\begin{bmatrix}1\\0\end{bmatrix},\quad
\widehat\nabla\widehat\psi_{\widehat K}^3=\begin{bmatrix}0\\1\end{bmatrix}.
$$
`psi0`, `psi1`, `psi2` correspondem, nessa ordem, aos índices $r=1,2,3$ da apostila. As listas `psi` e `grad_psi` permitem acessar essas funções pelos índices locais do programa. Nos elementos físicos,
$$
\psi_{K_m}^r=\widehat\psi_{\widehat K}^r\circ F_{K_m}^{-1},\qquad
\nabla\psi_{K_m}^r=B_{K_m}^{-T}\widehat\nabla\widehat\psi_{\widehat K}^r.
$$
A última identidade é a transformação de gradientes empregada na equação (8.2).

```python
def u(x):
  return np.sin(2 * np.pi * x[0]) + 0.5 * np.cos(5 * np.pi * x[1])

def psi0(xi, eta):
    return 1 - xi - eta

def psi1(xi, eta):
    return xi

def psi2(xi, eta):
    return eta

psi = [psi0, psi1, psi2]

# 2. Definir os gradientes das funções de base no elemento de referência
def grad_psi0(xi, eta):
    return np.array([-1.0, -1.0])

def grad_psi1(xi, eta):
    return np.array([1.0, 0.0])

def grad_psi2(xi, eta):
    return np.array([0.0, 1.0])

grad_psi = [grad_psi0, grad_psi1, grad_psi2]
```

## 6. Montagem da projeção $L^2(\Omega)$ — itens 1 e 2

Para qualquer um dos espaços $W_h=X_h$ ou $W_h=V_h$, procura-se $\mathcal P_h^{L^2}(u)\in W_h$ tal que
$$
\int_\Omega \mathcal P_h^{L^2}(u)v_h\,d\Omega
=\int_\Omega uv_h\,d\Omega\qquad\forall v_h\in W_h.
$$
Com $u_h=\sum_J U_J\phi^J$, obtém-se $\mathbf M\mathbf U=\mathbf F$, onde
$$
M_{IJ}=\int_\Omega\phi^I\phi^J\,d\Omega,\qquad
F_I=\int_\Omega u\phi^I\,d\Omega.
$$
Esta é a **matriz de massa** da equação (8.1); não há produto de gradientes nesta projeção. Localmente,
$$
M_{rs}^{(m)}\approx |J_{K_m}|\sum_g w_g\widehat\psi^r(\widehat{\mathbf x}_g)\widehat\psi^s(\widehat{\mathbf x}_g),
\qquad
F_r^{(m)}\approx |J_{K_m}|\sum_g w_g u(F_{K_m}(\widehat{\mathbf x}_g))\widehat\psi^r(\widehat{\mathbf x}_g).
$$
O índice $\widehat K$ das bases foi suprimido apenas nessas fórmulas para facilitar a leitura. A rotina percorre elementos `m`, pontos `g` e índices locais `r,s`; acumula a matriz `ae` e o vetor `rhse`; e realiza a montagem global:
$$
I=\mathrm{conec}(K_m,r),\quad J=\mathrm{conec}(K_m,s),\qquad
\mathrm{Aglob}_{IJ}\mathrel{+}=\mathrm{ae}_{rs},\quad
\mathrm{RHS}_I\mathrel{+}=\mathrm{rhse}_r.
$$
`Aglob` corresponde a `Aglo` no pseudocódigo da p. 82; `RHS`, `ae` e `rhse` mantêm a mesma função. Nesta rotina `invBk_T` é calculada, mas não é utilizada, pois a projeção $L^2$ não envolve derivadas.

As fórmulas acima usam $w_g$ da apostila; as operações do código usam $\widetilde w_g=2w_g$, conforme a seção de quadratura. Além disso, `np.zeros((nnodes, nnodes))` cria uma **matriz densa**. Embora a estrutura matemática seja esparsa, a exigência de armazenamento em formato esparso do item 1 não é atendida nesta versão. Nenhuma modificação foi feita para ocultar ou corrigir essa diferença.

```python
def montar_sistema_L2(nodes, elements, u, psi, ordem_quadratura="cubic"):
    """
    Monta a matriz global Aglob e o vetor de carga RHS para elementos finitos em 2D.

    Parâmetros:
    -----------
    nodes : ndarray
        Coordenadas dos nós físicos da malha.
    elements : ndarray
        Conectividade dos elementos da malha.
    u : function
        Função analítica u(x, y) do lado direito.
    psi : list of functions
        Funções de base locais no elemento de referência.
    ordem_quadratura : str, opcional
        Ordem da regra de integração de Gauss (padrão: "cubic").

    Retorna:
    --------
    Aglob : ndarray
        Matriz de rigidez/massa global.
    RHS : ndarray
        Vetor de força/carga global.
    """
    # 1. Carregar pontos e pesos de integração no triângulo de referência
    pontos_gauss, pesos = pontos_pesos_triangulo(ordem_quadratura)

    nloc = len(psi)
    nnodes = len(nodes)

    # 2. Inicializar matriz e vetor globais
    Aglob = np.zeros((nnodes, nnodes))
    RHS = np.zeros(nnodes)#ndofs

    # 3. Laço sobre cada elemento (triângulo) da malha
    for m in range(len(elements)):
        v0 = nodes[elements[m, 0]]
        Bk = getBk(element=elements[m], nodes=nodes)
        invBk_T = np.linalg.inv(np.transpose(Bk))
        detJac = np.abs(np.linalg.det(Bk))

        ae = np.zeros((nloc, nloc))
        rhse = np.zeros(nloc)

        # Integração numérica
        for g in range(len(pesos)):
            xi = pontos_gauss[g][0]
            eta = pontos_gauss[g][1]

            # Mapeamento para o domínio físico
            x_fisico = Bk @ np.array([xi, eta]) + v0

            for r in range(nloc):
                rhse[r] += detJac * pesos[g] * psi[r](xi, eta) * u(x_fisico)

                for s in range(nloc):
                    ae[r, s] += detJac * pesos[g] * psi[r](xi, eta) * psi[s](xi, eta)

        # Montagem (espalhamento) no sistema global
        for r in range(nloc):
            I = elements[m, r]
            RHS[I] += rhse[r]
            for s in range(nloc):
                J = elements[m, s]
                Aglob[I, J] += ae[r, s]

    return Aglob, RHS
```

## 7. Montagem da projeção $H^1(\Omega)$ — itens 1 e 2

Como na seção 8.3, procura-se $\mathcal P_h^{H^1}(u)\in V_h$ satisfazendo
$$
\int_\Omega\bigl(\mathcal P_h^{H^1}(u)v_h+\nabla\mathcal P_h^{H^1}(u)\cdot\nabla v_h\bigr)\,d\Omega
=\int_\Omega(uv_h+\nabla u\cdot\nabla v_h)\,d\Omega\quad\forall v_h\in V_h.
$$
Assim, $\mathbf A=\mathbf M+\mathbf K$, com $\mathbf M$ e $\mathbf K$ das equações (8.1) e (8.2). Definindo $\mathbf G_r=B_{K_m}^{-T}\widehat\nabla\widehat\psi^r$, as contribuições são
$$
A_{rs}^{(m)}\approx |J_{K_m}|\sum_g w_g\bigl(\widehat\psi^r\widehat\psi^s+\mathbf G_r\cdot\mathbf G_s\bigr),
$$
$$
F_r^{(m)}\approx |J_{K_m}|\sum_g w_g\bigl(u(F_{K_m}(\widehat{\mathbf x}_g))\widehat\psi^r+\nabla u(F_{K_m}(\widehat{\mathbf x}_g))\cdot\mathbf G_r\bigr).
$$
As bases são avaliadas no ponto $\widehat{\mathbf x}_g$. `grad_psi_fisico` guarda os vetores $\mathbf G_r$, e a montagem local-global segue o mesmo procedimento anterior. Vale a mesma observação sobre os pesos duplicados e o armazenamento denso.

**Por que somente o espaço contínuo?** $V_h\subset H^1(\Omega)$, mas $X_h$ admite saltos entre elementos e, em geral, não pertence a $H^1(\Omega)$. Somar apenas gradientes calculados no interior de cada elemento de uma função descontínua definiria um produto interno quebrado, e não a projeção $H^1(\Omega)$ solicitada. O programa usa `CG1` para esta projeção. Não se impõem condições de Dirichlet nas projeções; o termo de massa torna $\mathbf M+\mathbf K$ invertível mesmo incluindo funções constantes.

```python
def montar_sistema_H1(nodes, elements, u, grad_u, psi, grad_psi, ordem_quadratura="cubic"):
    """
    Monta a matriz global Aglob e o vetor de carga RHS para a projeção H1 em 2D.
    O produto interno em H1 é definido por: (u, v)_{H1} = (u, v)_{L2} + (grad(u), grad(v))_{L2}

    Parâmetros:
    -----------
    nodes : ndarray
        Coordenadas dos nós físicos da malha.
    elements : ndarray
        Conectividade dos elementos da malha.
    u : function
        Função analítica u(x, y).
    grad_u : function
        Gradiente da função analítica, retorna [du/dx, du/dy] dado um ponto físico [x, y].
    psi : list of functions
        Funções de base locais no elemento de referência.
    grad_psi : list of functions
        Gradientes das funções de base locais no elemento de referência,
        cada uma retorna [dpsi/dxi, dpsi/deta].
    ordem_quadratura : str, opcional
        Ordem da regra de integração de Gauss (padrão: "cubic").
    """
    # 1. Carregar pontos e pesos de integração no triângulo de referência
    pontos_gauss, pesos = pontos_pesos_triangulo(ordem_quadratura)

    nloc = len(psi)
    nnodes = len(nodes)

    # 2. Inicializar matriz e vetor globais
    Aglob = np.zeros((nnodes, nnodes))
    RHS = np.zeros(nnodes)

    # 3. Laço sobre cada elemento (triângulo) da malha
    for m in range(len(elements)):
        v0 = nodes[elements[m, 0]]
        Bk = getBk(element=elements[m], nodes=nodes)
        invBk_T = np.linalg.inv(np.transpose(Bk))
        detJac = np.abs(np.linalg.det(Bk))

        ae = np.zeros((nloc, nloc))
        rhse = np.zeros(nloc)

        # Integração numérica
        for g in range(len(pesos)):
            xi = pontos_gauss[g][0]
            eta = pontos_gauss[g][1]

            # Mapeamento para o domínio físico
            x_fisico = Bk @ np.array([xi, eta]) + v0

            # Avaliação dos valores de u e grad_u no ponto físico
            u_val = u(x_fisico)
            grad_u_val = np.array(grad_u(x_fisico))

            # Pré-computar os gradientes físicos das funções de base neste ponto de Gauss
            grad_psi_fisico = []
            for r in range(nloc):
                # Gradiente de referência [dpsi/dxi, dpsi/deta]
                g_ref = np.array(grad_psi[r](xi, eta))
                # Gradiente físico: J^-T * grad_ref
                g_fisico = invBk_T @ g_ref
                grad_psi_fisico.append(g_fisico)

            for r in range(nloc):
                psi_r_val = psi[r](xi, eta)

                # Termo fonte RHS em H1: u*v + grad(u).grad(v)
                termo_L2_rhs = psi_r_val * u_val
                termo_H1_rhs = np.dot(grad_psi_fisico[r], grad_u_val)

                rhse[r] += detJac * pesos[g] * (termo_L2_rhs + termo_H1_rhs)

                for s in range(nloc):
                    psi_s_val = psi[s](xi, eta)

                    # Matriz de Rigidez H1 bilinear: phi*psi + grad(phi).grad(psi)
                    termo_L2_mat = psi_r_val * psi_s_val
                    termo_H1_mat = np.dot(grad_psi_fisico[r], grad_psi_fisico[s])

                    ae[r, s] += detJac * pesos[g] * (termo_L2_mat + termo_H1_mat)

        # Montagem (espalhamento) no sistema global
        for r in range(nloc):
            I = elements[m, r]
            RHS[I] += rhse[r]
            for s in range(nloc):
                J = elements[m, s]
                Aglob[I, J] += ae[r, s]

    return Aglob, RHS
```

## 8. Execução das projeções $L^2$: contínua e descontínua — item 2

Este bloco instancia a malha com $n_x=n_y=5$ e $L_x=L_y=1$, resultando em $N_v=25$ e $N_{el}=32$. A quadratura cúbica é usada na montagem. Resolvem-se, sucessivamente,
$$
\mathbf M_{CG1}\mathbf U_{CG1}=\mathbf F_{CG1},\quad \dim V_h=25,
\qquad
\mathbf M_{DG1}\mathbf U_{DG1}=\mathbf F_{DG1},\quad \dim X_h=96.
$$
`uh_cg` e `uh_dg` guardam os coeficientes de $\mathcal P_h^{L^2}(u)$ nos dois espaços. `np.linalg.solve` resolve os sistemas densos. As saídas armazenadas confirmam matrizes $25\times25$ e $96\times96$.

A massa tem integrando quadrático e é integrada exatamente pela regra cúbica, após considerar a normalização. O vetor envolve funções trigonométricas; sua integração é aproximada. Portanto, trata-se de uma aproximação por quadratura da projeção definida pelas integrais exatas. Ao fim da célula, os nomes genéricos `nodes` e `elements` se referem a `DG1`, razão pela qual as visualizações seguintes recuperam explicitamente os dados de cada espaço.

## 12. Execução e comparação das projeções $L^2$ e $H^1$ em $CG1$ — item 2

Calcula-se analiticamente
$$
\nabla u(x,y)=\begin{bmatrix}2\pi\cos(2\pi x)\\-\frac52\pi\sin(5\pi y)\end{bmatrix}.
$$
A função `grad_u` fornece esse gradiente para o vetor da projeção $H^1$. A célula reutiliza `nodes_geom` e `triangulos` definidos anteriormente, configura `CG1`, monta $\mathbf M+\mathbf K$ com quadratura cúbica e resolve o sistema para `uh_h1`.

A comparação é entre $\mathcal P_h^{L^2}(u)$ e $\mathcal P_h^{H^1}(u)$ no mesmo espaço de dimensão $25$. Nas definições com integração exata, a primeira minimiza $\|u-v_h\|_{L^2}$, e a segunda minimiza $\|u-v_h\|_{H^1}$. Logo, a projeção $H^1$ considera simultaneamente valores e gradientes. Os painéis mostram essa diferença qualitativamente; a avaliação quantitativa vem nas próximas células. As soluções aqui calculadas usam integração numérica, que deve ser considerada ao interpretar essas propriedades.

## 13. Cálculo do erro em $L^2(\Omega)$ — item 3

Para cada elemento, a conectividade extrai `uh_local`, e as bases reconstruem $u_h$ nos pontos de quadratura. A norma desejada é
$$
E_{L^2}=\|u-u_h\|_{L^2(\Omega)}
=\left(\sum_m\int_{K_m}|u-u_h|^2\,dK\right)^{1/2}.
$$
Com os pesos da apostila, sua aproximação é
$$
E_{L^2,Q}=\left[\sum_m|J_{K_m}|\sum_g w_g
\left|u(F_{K_m}(\widehat{\mathbf x}_g))-
\sum_rU_{LG_{K_m}(r)}\widehat\psi^r(\widehat{\mathbf x}_g)\right|^2\right]^{1/2}.
$$
A rotina serve tanto para `CG1` quanto para `DG1`, pois ambos estão contidos em $L^2(\Omega)$. Como utiliza $\widetilde w_g=2w_g$, o valor retornado satisfaz $E_{\mathrm{codigo}}=\sqrt2\,E_{L^2,Q}$ quando a soma admite raiz real. Essa equivalência é relativa à mesma regra de quadratura; não elimina o erro de integração da função trigonométrica.

```python
def calcular_erro_L2(nodes, elements, u, uh, psi, ordem_quadratura="cubic"):
    """
    Calcula a norma L2 do erro da aproximação: ||u - u_h||_L2

    Parâmetros:
    -----------
    nodes : ndarray
        Coordenadas dos nós físicos da malha.
    elements : ndarray
        Conectividade dos elementos da malha.
    u : function
        Função analítica exata u(x, y).
    uh : ndarray
        Vetor com os coeficientes da solução aproximada nos nós.
    psi : list of functions
        Funções de base locais no elemento de referência.
    ordem_quadratura : str, opcional
        Ordem da regra de integração de Gauss (padrão: "cubic").

    Retorna:
    --------
    erro_L2 : float
        A norma L2 global do erro.
    """
    pontos_gauss, pesos = pontos_pesos_triangulo(ordem_quadratura)
    nloc = len(psi)
    erro_quadrado_total = 0.0

    # Laço sobre cada elemento da malha
    for m in range(len(elements)):
        v0 = nodes[elements[m, 0]]
        Bk = getBk(element=elements[m], nodes=nodes)
        detJac = np.abs(np.linalg.det(Bk))

        # Obter os coeficientes aproximados locais para o elemento m
        uh_local = uh[elements[m]]

        # Integração numérica sobre o elemento atual
        for g in range(len(pesos)):
            xi = pontos_gauss[g][0]
            eta = pontos_gauss[g][1]

            # Mapeamento do ponto de integração para o domínio físico
            x_fisico = Bk @ np.array([xi, eta]) + v0

            # Avaliar u exata no ponto físico
            u_exata_val = u(x_fisico)

            # Avaliar u aproximada (u_h) usando as funções de base
            uh_val = 0.0
            for r in range(nloc):
                uh_val += uh_local[r] * psi[r](xi, eta)

            # Diferença ao quadrado ponderada com o Jacobiano e peso
            erro_quadrado_total += detJac * pesos[g] * ((u_exata_val - uh_val) ** 2)

    return np.sqrt(erro_quadrado_total)
```

## 14. Cálculo do erro em $H^1(\Omega)$ — item 3

Para a aproximação contínua, calcula-se a norma completa, e não somente a seminorma do gradiente:
$$
E_{H^1}=\|u-u_h\|_{H^1(\Omega)}
=\left[\sum_m\int_{K_m}\left(|u-u_h|^2+|\nabla u-\nabla u_h|^2\right)dK\right]^{1/2},
$$
$$
\nabla u_h|_{K_m}=\sum_{r=1}^{3}U_{LG_{K_m}(r)}B_{K_m}^{-T}\widehat\nabla\widehat\psi_{\widehat K}^r.
$$
`uh_val` e `grad_uh_val` reconstroem o valor e o gradiente físicos; `termo_L2` e `termo_grad` são as duas parcelas do integrando. A função aplica quadratura e retorna a raiz da soma global.

Ela é chamada apenas para aproximações `CG1`. Aplicá-la a `DG1` somaria gradientes por elemento e não produziria a norma global $H^1(\Omega)$ de uma função com saltos. Pela normalização dos pesos, o resultado também é $\sqrt2$ vezes a aproximação de norma obtida com os pesos da apostila.

```python
def calcular_erro_H1(nodes, elements, u, grad_u, uh, psi, grad_psi, ordem_quadratura="cubic"):
    """
    Calcula a norma H1 do erro da aproximação: ||u - u_h||_H1

    Parâmetros:
    -----------
    nodes : ndarray
        Coordenadas dos nós físicos da malha.
    elements : ndarray
        Conectividade dos elementos da malha.
    u : function
        Função analítica exata u(x, y).
    grad_u : function
        Gradiente da função analítica exata, retorna [du/dx, du/dy].
    uh : ndarray
        Vetor com os coeficientes da solução aproximada nos nós.
    psi : list of functions
        Funções de base locais no elemento de referência.
    grad_psi : list of functions
        Gradientes das funções de base locais no elemento de referência.
    ordem_quadratura : str, opcional
        Ordem da regra de integração de Gauss (padrão: "cubic").
    """
    pontos_gauss, pesos = pontos_pesos_triangulo(ordem_quadratura)
    nloc = len(psi)
    erro_H1_quadrado = 0.0

    for m in range(len(elements)):
        v0 = nodes[elements[m, 0]]
        Bk = getBk(element=elements[m], nodes=nodes)
        invBk_T = np.linalg.inv(np.transpose(Bk))
        detJac = np.abs(np.linalg.det(Bk))

        uh_local = uh[elements[m]]

        for g in range(len(pesos)):
            xi = pontos_gauss[g][0]
            eta = pontos_gauss[g][1]

            x_fisico = Bk @ np.array([xi, eta]) + v0

            u_exata_val = u(x_fisico)
            grad_u_exata = np.array(grad_u(x_fisico))

            uh_val = 0.0
            grad_uh_val = np.zeros(2)

            for r in range(nloc):
                psi_val = psi[r](xi, eta)
                g_ref = np.array(grad_psi[r](xi, eta))
                g_fisico = invBk_T @ g_ref

                uh_val += uh_local[r] * psi_val
                grad_uh_val += uh_local[r] * g_fisico

            termo_L2 = (u_exata_val - uh_val) ** 2
            termo_grad = np.sum((grad_u_exata - grad_uh_val) ** 2)

            erro_H1_quadrado += detJac * pesos[g] * (termo_L2 + termo_grad)

    return np.sqrt(erro_H1_quadrado)
```

## 15. Comparação dos erros na malha inicial — item 3

A regra quíntica avalia quatro quantidades: erro $L^2$ da projeção $L^2$ em `CG1`; erro $L^2$ da projeção $L^2$ em `DG1`; erro $H^1$ da projeção $L^2$ em `CG1`; e erro $H^1$ da projeção $H^1$ em `CG1`. São **três projeções e quatro medidas de erro**, e não quatro problemas de contorno.

As saídas já armazenadas no notebook são, nessa ordem,
$$
4.727360\times10^{-1},\quad2.189473\times10^{-1},\quad
7.512472,\quad7.313284.
$$
Esses valores foram preservados e não representam uma nova execução durante a redação. Devem ser interpretados com o fator $\sqrt2$ explicado acima. O menor erro $L^2$ de `DG1` é compatível com sua maior liberdade de aproximação; o menor erro $H^1$ de `uh_h1` é compatível com o produto interno usado na projeção. As montagens iniciais usaram regra cúbica, enquanto esta avaliação usa regra quíntica.

```python
# Recalculando erros na norma L2
erro_l2_cg = calcular_erro_L2(nodes_cg, elements_cg, u, uh_cg, psi, ordem_quadratura="quintic")
erro_l2_dg = calcular_erro_L2(nodes_dg, elements_dg, u, uh_dg, psi, ordem_quadratura="quintic")

# Calculando erros na norma H1
erro_h1_cg = calcular_erro_H1(nodes_cg, elements_cg, u, grad_u, uh_cg, psi, grad_psi, ordem_quadratura="quintic")
erro_h1_h1_proj = calcular_erro_H1(nodes_h1, elements_h1, u, grad_u, uh_h1, psi, grad_psi, ordem_quadratura="quintic")

print("=================== ERROS DE APROXIMAÇÃO ===================\n")
print(f"Norma L2 - Espaço CG1: {erro_l2_cg:.6e}")
print(f"Norma L2 - Espaço DG1: {erro_l2_dg:.6e}\n")
print(f"Norma H1 - Projeção L2 em CG1: {erro_h1_cg:.6e}")
print(f"Norma H1 - Projeção H1 em CG1: {erro_h1_h1_proj:.6e}")
print("============================================================")
```

## 16. Refinamento da malha e taxas de convergência — item 3

Agora as três projeções são recalculadas para `N_valores = [4,8,16,32]`, com quadratura quíntica tanto na montagem quanto no cálculo dos erros. Nesta célula, `N` significa o número de pontos em **cada direção**, não $\dim V_h$:
$$
h_{\mathrm{codigo}}=\frac1{N-1},\qquad N_{el}=2(N-1)^2,
\qquad\dim V_h=N^2,\qquad\dim X_h=6(N-1)^2.
$$
A definição de tamanho pelo maior diâmetro dos triângulos dá $h_{\mathcal T}=\sqrt2\,h_{\mathrm{codigo}}$ nesta malha. O fator constante não altera as taxas. Para duas malhas consecutivas, o programa calcula
$$
p_k=\frac{\log(E_k/E_{k-1})}{\log(h_k/h_{k-1})}.
$$
O primeiro valor é `nan`, pois ainda não há duas malhas para comparação. O fator $\sqrt2$ nos erros também se cancela nessa razão, embora altere seus valores absolutos.

Para funções suaves e elementos $P_1$ em malhas regulares, esperam-se comportamentos de ordem $h^2$ no erro $L^2$ da projeção $L^2$ e de ordem $h$ no erro $H^1$ das aproximações contínuas aqui comparadas. A quadratura e malhas ainda grossas podem afastar as taxas desse regime.

Na tabela previamente armazenada, as taxas finais são $2.12$ (`CG1`, erro $L^2$), $1.99$ (`DG1`, erro $L^2$), $1.02$ (projeção $L^2$, erro $H^1$) e $0.98$ (projeção $H^1$, erro $H^1$). Esses dados são compatíveis com as ordens esperadas; não constituem prova de convergência. As listas e a tabela são produzidas exatamente pelo código abaixo, sem alterações no solver.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Definição das partições de malha para a análise de convergência
N_valores = [4, 8, 16, 32]

# Listas para armazenar os erros obtidos
erros_L2_cg = []
erros_L2_dg = []
erros_H1_cg_projL2 = []
erros_H1_cg_projH1 = []

h_valores = []

print("Iniciando simulações para análise de convergência...")

for N in N_valores:
    h = 1.0 / (N - 1)
    h_valores.append(h)
    print(f"Processando malha com N = {N} (h = {h:.4f})...")

    # 1. Gerar malha geométrica básica
    nodes_g, tri_g = gerar_malha_2d(lx=1.0, ly=1.0, nx=N, ny=N)

    # --- Espaço CG1 ---
    nodes_cg, elements_cg = configurar_espaco_aproximacao(nodes_g, tri_g, type_space="CG1")
    # Projeção L2
    A_cg, b_cg = montar_sistema_L2(nodes_cg, elements_cg, u, psi, ordem_quadratura="quintic")
    uh_cg_l2 = np.linalg.solve(A_cg, b_cg)
    # Projeção H1
    A_cg_h1, b_cg_h1 = montar_sistema_H1(nodes_cg, elements_cg, u, grad_u, psi, grad_psi, ordem_quadratura="quintic")
    uh_cg_h1 = np.linalg.solve(A_cg_h1, b_cg_h1)

    # --- Espaço DG1 ---
    nodes_dg, elements_dg = configurar_espaco_aproximacao(nodes_g, tri_g, type_space="DG1")
    # Projeção L2
    A_dg, b_dg = montar_sistema_L2(nodes_dg, elements_dg, u, psi, ordem_quadratura="quintic")
    uh_dg_l2 = np.linalg.solve(A_dg, b_dg)

    # 2. Calcular Erros com quadratura quintic
    err_l2_cg = calcular_erro_L2(nodes_cg, elements_cg, u, uh_cg_l2, psi, ordem_quadratura="quintic")
    err_l2_dg = calcular_erro_L2(nodes_dg, elements_dg, u, uh_dg_l2, psi, ordem_quadratura="quintic")

    err_h1_cg_projL2 = calcular_erro_H1(nodes_cg, elements_cg, u, grad_u, uh_cg_l2, psi, grad_psi, ordem_quadratura="quintic")
    err_h1_cg_projH1 = calcular_erro_H1(nodes_cg, elements_cg, u, grad_u, uh_cg_h1, psi, grad_psi, ordem_quadratura="quintic")

    # Armazenar resultados
    erros_L2_cg.append(err_l2_cg)
    erros_L2_dg.append(err_l2_dg)
    erros_H1_cg_projL2.append(err_h1_cg_projL2)
    erros_H1_cg_projH1.append(err_h1_cg_projH1)

# Função auxiliar para calcular taxas de convergência
def calcular_taxas(h_list, erro_list):
    taxas = [np.nan]
    for i in range(1, len(h_list)):
        rate = np.log(erro_list[i] / erro_list[i-1]) / np.log(h_list[i] / h_list[i-1])
        taxas.append(rate)
    return taxas

taxas_L2_cg = calcular_taxas(h_valores, erros_L2_cg)
taxas_L2_dg = calcular_taxas(h_valores, erros_L2_dg)
taxas_H1_cg_l2 = calcular_taxas(h_valores, erros_H1_cg_projL2)
taxas_H1_cg_h1 = calcular_taxas(h_valores, erros_H1_cg_projH1)

# 3. Construir Tabela de Erros
data = {
    'N': N_valores,
    'h': h_valores,
    'Erro L2 (CG1)': erros_L2_cg,
    'Taxa L2 (CG1)': taxas_L2_cg,
    'Erro L2 (DG1)': erros_L2_dg,
    'Taxa L2 (DG1)': taxas_L2_dg,
    'Erro H1 (Proj L2)': erros_H1_cg_projL2,
    'Taxa H1 (Proj L2)': taxas_H1_cg_l2,
    'Erro H1 (Proj H1)': erros_H1_cg_projH1,
    'Taxa H1 (Proj H1)': taxas_H1_cg_h1
}
df_erros = pd.DataFrame(data)
display(df_erros.style.format({
    'h': '{:.4e}',
    'Erro L2 (CG1)': '{:.6e}', 'Taxa L2 (CG1)': '{:.2f}',
    'Erro L2 (DG1)': '{:.6e}', 'Taxa L2 (DG1)': '{:.2f}',
    'Erro H1 (Proj L2)': '{:.6e}', 'Taxa H1 (Proj L2)': '{:.2f}',
    'Erro H1 (Proj H1)': '{:.6e}', 'Taxa H1 (Proj H1)': '{:.2f}'
}))
```

## 18. Problema de difusão-reação e definição dos três casos — item 4

Passamos das projeções de uma função conhecida à determinação de uma solução desconhecida. O problema da seção 8.5 é
$$
\begin{cases}
-\nabla\cdot(\mu(\mathbf x)\nabla u(\mathbf x))+\sigma(\mathbf x)u(\mathbf x)=f(\mathbf x),&\mathbf x\in\Omega,\\
u(\mathbf x)=g(\mathbf x),&\mathbf x\in\Gamma_D,\\
-\mu(\mathbf x)\nabla u(\mathbf x)\cdot\check{\mathbf n}=0,&\mathbf x\in\Gamma_N,
\end{cases}
$$
com $\check{\mathbf n}$ a normal exterior. As partes da fronteira são indicadas pelos mesmos nomes da apostila:
$$
\Gamma_{\mathrm{left}}=\{0\}\times(0,1),\quad
\Gamma_{\mathrm{right}}=\{1\}\times(0,1),\quad
\Gamma_{\mathrm{bottom}}=(0,1)\times\{0\},\quad
\Gamma_{\mathrm{top}}=(0,1)\times\{1\}.
$$
A classificação dos vértices é feita por `is_dirichlet` usando `np.isclose`. Vértices de uma aresta de Dirichlet recebem o valor prescrito, inclusive nos encontros com arestas de Neumann.

### Caso 1 — reação-difusão com Neumann homogênea em toda a fronteira
$$
\mu=\sigma=f=1,\qquad\Gamma_D=\varnothing,\qquad\Gamma_N=\partial\Omega.
$$
O problema é $-\Delta u+u=1$, com $-\nabla u\cdot\check{\mathbf n}=0$. `is_dirichlet` retorna sempre `False`, e o valor de `g` não é utilizado. A solução exata é $u\equiv1$. O termo de reação $\sigma=1$ garante unicidade: este não é o problema de Poisson puro com Neumann e indeterminação por constantes. Como constantes pertencem a `CG1`, este caso oferece uma verificação natural da montagem.

### Caso 2 — reação-difusão com Dirichlet homogênea em toda a fronteira
$$
\mu=\sigma=f=1,\qquad\Gamma_N=\varnothing,\qquad\Gamma_D=\partial\Omega,\qquad g=0.
$$
Mantém-se $-\Delta u+u=1$, mas agora todos os graus de liberdade da fronteira são prescritos como zero. O campo interior responde à fonte e deve ser distinguido do caso anterior, apesar dos coeficientes iguais. A célula final apresenta sua solução numérica; não fornece uma solução analítica para comparação de erros.

### Caso 3 — coeficientes variáveis e condições mistas
$$
\mu(x,y)=1+x^2+y^2,\qquad\sigma(x,y)=xy,\qquad f(x,y)=x+y,
$$
$$
\Gamma_D=\Gamma_{\mathrm{left}}\cup\Gamma_{\mathrm{bottom}},\qquad
\Gamma_N=\Gamma_{\mathrm{right}}\cup\Gamma_{\mathrm{top}},
$$
$$
g(x,y)=\begin{cases}y(1-y),&(x,y)\in\Gamma_{\mathrm{left}},\\x(1-x),&(x,y)\in\Gamma_{\mathrm{bottom}}.\end{cases}
$$
As expressões de $g$ coincidem em $(0,0)$, onde valem zero. No código, `g_func` testa primeiro a esquerda e depois a base, de acordo com esses dados. A difusão é estritamente positiva e a reação é não negativa no domínio. As fronteiras direita e superior têm fluxo normal homogêneo.

A função `obter_parametros_caso` apenas retorna $\mu,\sigma,f,g$ e o identificador da fronteira de Dirichlet. A função trigonométrica `u` usada nas projeções **não é** uma solução exata prescrita para estes casos.

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Definição dos parâmetros para os Três Casos ---

def obter_parametros_caso(caso):
    if caso == 1:
        # Caso 1: mu = sigma = f = 1, Dirichlet vazia (apenas Neumann homogênea)
        mu = lambda x: 1.0
        sigma = lambda x: 1.0
        f = lambda x: 1.0
        # Condição Dirichlet (vazia)
        is_dirichlet = lambda x: False
        g = lambda x: 0.0

    elif caso == 2:
        # Caso 2: mu = sigma = f = 1, Neumann vazia (apenas Dirichlet homogênea g = 0)
        mu = lambda x: 1.0
        sigma = lambda x: 1.0
        f = lambda x: 1.0
        # Fronteira de Dirichlet é todo o contorno
        is_dirichlet = lambda x: (np.isclose(x[0], 0.0) or np.isclose(x[0], 1.0) or
                                  np.isclose(x[1], 0.0) or np.isclose(x[1], 1.0))
        g = lambda x: 0.0

    elif caso == 3:
        # Caso 3: mu, sigma e f variáveis.
        # Dirichlet em x=0 (esquerda) e y=0 (base). Neumann em x=1 (direita) e y=1 (topo)
        mu = lambda x: 1.0 + x[0]**2 + x[1]**2
        sigma = lambda x: x[0] * x[1]
        f = lambda x: x[0] + x[1]

        is_dirichlet = lambda x: (np.isclose(x[0], 0.0) or np.isclose(x[1], 0.0))

        def g_func(x):
            if np.isclose(x[0], 0.0): # Esquerda (Gamma_left)
                return x[1] * (1.0 - x[1])
            elif np.isclose(x[1], 0.0): # Base (Gamma_bottom)
                return x[0] * (1.0 - x[0])
            return 0.0
        g = g_func
    else:
        raise ValueError("Caso inválido.")

    return mu, sigma, f, is_dirichlet, g
```

## 19. Formulação variacional e montagem de difusão-reação — item 4

Definimos $V_{h,0}=\{v_h\in V_h:v_h=0\text{ em }\Gamma_D\}$ e o conjunto afim $V_{h,g_h}$ com os valores nodais de $g$ prescritos na fronteira. Busca-se $u_h\in V_{h,g_h}$ tal que
$$
a(u_h,v_h)=\ell(v_h)\quad\forall v_h\in V_{h,0},
$$
$$
a(u_h,v_h)=\int_\Omega\bigl(\mu\nabla u_h\cdot\nabla v_h+\sigma u_hv_h\bigr)d\Omega,
\qquad\ell(v_h)=\int_\Omega fv_h\,d\Omega.
$$
A integração por partes gera um termo de fronteira envolvendo o fluxo normal. Ele é nulo em $\Gamma_N$ pela condição de Neumann homogênea e em $\Gamma_D$ porque $v_h=0$. Portanto, não há integral de fronteira a acrescentar ao vetor neste exercício. Para o caso 1, $V_{h,0}=V_h$ e não há valores essenciais prescritos.

No elemento $K_m$, com $\mathbf G_r=B_{K_m}^{-T}\widehat\nabla\widehat\psi^r$ e $\mathbf x_g=F_{K_m}(\widehat{\mathbf x}_g)$,
$$
A_{rs}^{(m)}\approx |J_{K_m}|\sum_g w_g\left[\mu(\mathbf x_g)\mathbf G_r\cdot\mathbf G_s+
\sigma(\mathbf x_g)\widehat\psi^r(\widehat{\mathbf x}_g)\widehat\psi^s(\widehat{\mathbf x}_g)\right],
$$
$$
F_r^{(m)}\approx |J_{K_m}|\sum_g w_g f(\mathbf x_g)\widehat\psi^r(\widehat{\mathbf x}_g).
$$
`mu_f`, `sigma_f` e `f_f` correspondem a $\mu$, $\sigma$ e $f$; `termo_dif` e `termo_reac` são as parcelas de difusão e reação. O espalhamento por `elements` é o mesmo da seção 8.3. A montagem ainda não impõe Dirichlet.

Mantêm-se os pesos normalizados para soma um e o armazenamento denso. Na chamada final usa-se regra quíntica: os integrandos dos casos 1 e 2 têm grau no máximo dois; no caso 3, o termo de difusão tem grau dois, o de reação grau quatro e o vetor grau dois. Assim, a regra de grau cinco é suficiente para esses integrandos, à precisão dos coeficientes tabulados e à normalização já discutida.

```python
def montar_sistema_geral(nodes, elements, mu_f, sigma_f, f_f, psi, grad_psi, ordem_quadratura="cubic"):
    """
    Monta a matriz global Aglob e o RHS para a equação geral de difusão-reação.
    """
    pontos_gauss, pesos = pontos_pesos_triangulo(ordem_quadratura)
    nloc = len(psi)
    nnodes = len(nodes)

    Aglob = np.zeros((nnodes, nnodes))
    RHS = np.zeros(nnodes)

    for m in range(len(elements)):
        v0 = nodes[elements[m, 0]]
        Bk = getBk(element=elements[m], nodes=nodes)
        invBk_T = np.linalg.inv(np.transpose(Bk))
        detJac = np.abs(np.linalg.det(Bk))

        ae = np.zeros((nloc, nloc))
        rhse = np.zeros(nloc)

        for g_idx in range(len(pesos)):
            xi = pontos_gauss[g_idx][0]
            eta = pontos_gauss[g_idx][1]
            x_fisico = Bk @ np.array([xi, eta]) + v0

            # Avaliação dos coeficientes no ponto físico
            mu_val = mu_f(x_fisico)
            sigma_val = sigma_f(x_fisico)
            f_val = f_f(x_fisico)

            for r in range(nloc):
                psi_r_val = psi[r](xi, eta)
                g_ref_r = np.array(grad_psi[r](xi, eta))
                g_fisico_r = invBk_T @ g_ref_r

                rhse[r] += detJac * pesos[g_idx] * psi_r_val * f_val

                for s in range(nloc):
                    psi_s_val = psi[s](xi, eta)
                    g_ref_s = np.array(grad_psi[s](xi, eta))
                    g_fisico_s = invBk_T @ g_ref_s

                    # Termo de difusão + Termo de reação
                    termo_dif = mu_val * np.dot(g_fisico_r, g_fisico_s)
                    termo_reac = sigma_val * psi_r_val * psi_s_val

                    ae[r, s] += detJac * pesos[g_idx] * (termo_dif + termo_reac)

        for r in range(nloc):
            I = elements[m, r]
            RHS[I] += rhse[r]
            for s in range(nloc):
                J = elements[m, s]
                Aglob[I, J] += ae[r, s]

    return Aglob, RHS
```

## 20. Imposição das condições de Dirichlet — item 4

Para cada grau de liberdade de fronteira $I$ identificado por `is_dirichlet`, a rotina substitui sua equação por
$$
U_I=g(\mathbf x_I),\qquad
(A_{\mathrm{cond}})_{IJ}=\delta_{IJ},\qquad(b_{\mathrm{cond}})_I=g(\mathbf x_I).
$$
São modificadas somente as linhas correspondentes, conservando as colunas e as equações dos graus de liberdade livres. Separando índices livres $F$ e prescritos $D$, as linhas livres continuam representando
$$
A_{FF}\mathbf U_F+A_{FD}\mathbf g_D=\mathbf b_F,
\qquad\text{equivalentemente}\qquad
A_{FF}\mathbf U_F=\mathbf b_F-A_{FD}\mathbf g_D.
$$
Logo, o efeito dos dados não homogêneos está incluído ao resolver o sistema completo, mesmo sem subtrair explicitamente $A_{FD}\mathbf g_D$ do vetor. A substituição somente de linhas pode perder a simetria da matriz; `np.linalg.solve` não exige simetria.

No caso 1, nenhuma linha é alterada; no caso 2, todas as linhas de fronteira são substituídas; no caso 3, apenas as da esquerda e da base. Como $g$ é quadrático nas arestas e o espaço é $P_1$, a condição discreta usa sua interpolação nodal $g_h$, linear por trecho, e não coincide com $g$ em todo ponto da aresta. O fator comum dois na matriz e no vetor das linhas livres continua se cancelando mesmo após a prescrição das linhas de Dirichlet.

```python
def aplicar_dirichlet(Aglob, RHS, nodes, is_dirichlet, g_f):
    """
    Aplica as condições de contorno de Dirichlet modificando a matriz e o RHS.
    """
    A_cond = Aglob.copy()
    b_cond = RHS.copy()

    for idx, node in enumerate(nodes):
        if is_dirichlet(node):
            val_g = g_f(node)
            # Zera a linha correspondente na matriz e coloca 1 na diagonal
            A_cond[idx, :] = 0.0
            A_cond[idx, idx] = 1.0
            b_cond[idx] = val_g

    return A_cond, b_cond
```

## 21. Resolução e visualização dos três casos — item 4

O bloco final gera uma nova malha com $n_x=n_y=30$:

N_v=900,\qquad N_{el}=2\cdot29^2=1682,\qquad\dim V_h=900.

Usa-se `CG1` para todos os casos. Para cada valor de `caso`, recuperam-se os dados, monta-se o sistema com quadratura quíntica, impõem-se os valores essenciais e resolve-se

O dicionário `solucoes` guarda separadamente os vetores dos casos 1, 2 e 3. As primeiras figuras mostram as superfícies 3D; as seguintes mostram curvas e regiões de nível de cada solução. Essas são referências para interpretar os gráficos, não medições adicionais de erro: esta célula não calcula resíduos nem erros dos problemas do item 4.

O notebook documenta, assim, as três projeções solicitadas, suas medidas de erro e os três casos de difusão-reação.
```python
# Resolução dos Três Casos em uma malha refinada
Nx, Ny = 30, 30
nodes_geom, triangulos = gerar_malha_2d(1.0, 1.0, Nx, Ny)
nodes_cg, elements_cg = configurar_espaco_aproximacao(nodes_geom, triangulos, type_space="CG1")

solucoes = {}

# 1. Resolver cada um dos três casos
for caso in [1, 2, 3]:
    mu, sigma, f, is_dirichlet, g = obter_parametros_caso(caso)
    A_raw, b_raw = montar_sistema_geral(nodes_cg, elements_cg, mu, sigma, f, psi, grad_psi, "quintic")
    A_sys, b_sys = aplicar_dirichlet(A_raw, b_raw, nodes_cg, is_dirichlet, g)
    uh = np.linalg.solve(A_sys, b_sys)
    solucoes[caso] = uh

# --- Visualização 1: Superfícies 3D ---
fig1 = plt.figure(figsize=(20, 5))
for caso in [1, 2, 3]:
    ax = fig1.add_subplot(1, 3, caso, projection='3d')
    surf = ax.plot_trisurf(nodes_cg[:, 0], nodes_cg[:, 1], solucoes[caso], triangles=elements_cg, cmap='coolwarm_r', edgecolor='none')
    ax.set_title(f'Solução do Caso {caso} (3D)', fontsize=12, fontweight='bold')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    fig1.colorbar(surf, ax=ax, shrink=0.5, aspect=10)

plt.suptitle('Comparação das Soluções dos Três Casos (Visualização 3D)', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# --- Visualização 2: Contornos 2D ---
fig2, axes2 = plt.subplots(1, 3, figsize=(20, 5))
for caso in [1, 2, 3]:
    ax = axes2[caso-1]
    levels = 25

    cf = ax.tricontourf(nodes_cg[:, 0], nodes_cg[:, 1], solucoes[caso], levels=levels, cmap='coolwarm_r')
    cl = ax.tricontour(nodes_cg[:, 0], nodes_cg[:, 1], solucoes[caso], levels=levels, colors='black', linewidths=0.3, alpha=0.5)

    ax.set_title(f'Caso {caso}: Contornos de Magnitude', fontsize=12, fontweight='bold')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_aspect('equal')
    fig2.colorbar(cf, ax=ax, shrink=0.8)

plt.suptitle('Campos de Contorno de Magnitude para os Três Casos', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
